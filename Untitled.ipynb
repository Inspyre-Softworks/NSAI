```python
import json
import re
import sqlite3

from nsai.nations import advice_cache_path


nation = "Free Separatists of New Columbia"


def website_reason(reason, option_number_by_id):
    """Convert cached Option IDs to their displayed website positions."""

    def convert_option(match):
        raw_id = match.group(1)

        if raw_id == "-1":
            return "dismissed"

        website_number = option_number_by_id.get(raw_id)

        if website_number is None:
            # Leave unknown references visible instead of misnumbering them.
            return f"Option {raw_id}"

        return f"Option {website_number}"

    reason = re.sub(
        r"\bOption\s+(-?\d+)\b",
        convert_option,
        str(reason),
        flags=re.IGNORECASE,
    )

    # Advisor-only recommendations have not necessarily been enacted.
    reason = re.sub(
        r"\bthe enacted option\b",
        "the chosen option",
        reason,
        flags=re.IGNORECASE,
    )

    return reason


db = sqlite3.connect(advice_cache_path())
db.row_factory = sqlite3.Row

try:
    # Use the active issue set from the most recent all-issues advisor run.
    plan = db.execute(
        """
        SELECT issue_ids_json, ordered_issue_ids_json, updated_at
        FROM issue_plans
        WHERE nation = ?
        ORDER BY updated_at DESC
        LIMIT 1
        """,
        (nation,),
    ).fetchone()

    if plan is None:
        print(f"No cached issue plan found for {nation}.")

    else:
        active_ids = {
            str(issue_id)
            for issue_id in json.loads(plan["issue_ids_json"])
        }

        ordered_ids = [
            str(issue_id)
            for issue_id in json.loads(plan["ordered_issue_ids_json"])
        ]

        # Preserve advisor priority order and include any omitted active issues.
        issue_ids = [
            *[
                issue_id
                for issue_id in ordered_ids
                if issue_id in active_ids
            ],
            *sorted(active_ids - set(ordered_ids)),
        ]

        print(f"Nation: {nation}")
        print(f"Advisor snapshot: {plan['updated_at']}")
        print()

        for issue_id in issue_ids:
            row = db.execute(
                """
                SELECT live_issue_json, recommendation_json
                FROM issue_advice
                WHERE nation = ? AND issue_id = ?
                """,
                (nation, issue_id),
            ).fetchone()

            if row is None:
                print(f"Issue {issue_id}")
                print("  Choice: no cached recommendation")
                print()
                continue

            issue = json.loads(row["live_issue_json"])
            recommendation = json.loads(row["recommendation_json"])

            # NationStates IDs may contain gaps. The website numbers options
            # according to their position in the displayed option list.
            option_number_by_id = {
                str(option.get("option_id")): position
                for position, option in enumerate(
                    issue.get("options", []),
                    start=1,
                )
            }

            action = str(
                recommendation.get("action", "")
            ).strip().lower()

            raw_option_id = str(
                recommendation.get("option_id", "")
            ).strip()

            if action == "dismiss" or raw_option_id == "-1":
                choice = "dismissed"
            else:
                website_number = option_number_by_id.get(raw_option_id)

                if website_number is None:
                    choice = f"unknown option ID {raw_option_id!r}"
                else:
                    choice = f"Option {website_number}"

            reason = website_reason(
                recommendation.get("reasoning")
                or recommendation.get("why_this_issue_first")
                or "(no reason cached)",
                option_number_by_id,
            )

            title = issue.get("title") or "Untitled issue"

            print(f"{title} (Issue {issue_id})")
            print(f"  Choice: {choice}")
            print(f"  Reason: {reason}")
            print()

finally:
    db.close()
```
